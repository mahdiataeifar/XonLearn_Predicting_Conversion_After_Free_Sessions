## مقدمه


<div class="cell rtl-align" dir="rtl" lang="fa" id="مقدمه">
فناوری های آموزشی (Educational technologies) به فناوری هایی گفته می شود که برای ارتقاء و تسهیل آموزش و یادگیری مورد استفاده قرار می گیرند؛ مانند: پلتفرم های یادگیری آنلاین، سیستم های مدیریت یادگیری، دوره های آنلاین و کتاب های الکترونیک.
در یک دهه گذشته، فناوری های آموزشی به سرعت رشد کرده اند. برخی از دلایل این موضوع عبارتند از:
<br></br>
* پیشرفت های فناورانه: نوآوری ها در هوش مصنوعی و محاسبات ابری، تجربیات یادگیری فراگیرتر و شخصی سازی شده بیشتری را ممکن کرده است.
<br></br>
* افزایش تقاضا برای آموزش آنلاین: همه گیری کووید به پذیرش راه حل های یادگیری آنلاین سرعت بخشیده است.
<br></br>
* سرمایه گذاری دولت ها: بسیاری از دولت ها از ترویج آموزش و یادگیری دیجیتال حمایت می کنند.
<br></br>
* یادگیری مادام العمر: سرعت سریع تغییرات تکنولوژیک مستلزم توسعه مداوم مهارت است که منجر به افزایش تقاضا برای دوره های آنلاین و برنامه های آموزشی شده است.
<br></br>
<div class="cell rtl-align" dir="rtl" lang="fa" id="شرح تمرین">
در میان انواع محصولات و خدمات آموزشی، آموزش آنلاین در سال های اخیر شاهد رشد سریعی بوده است و شرکت های بسیاری در این عرصه پا به ظهور گذاشته اند. این شرکت ها در تلاش هستند تا مخاطبان بیشتری را به سمت دوره های آموزشی خود جذب و آنان را به مشتریان خود بدل کنند.
<br></br>
    افرادی که به خدمات و محصولات شرکت ها علاقه نشان می دهند، مشتریان  بالقوه (potential customer) نامیده می شوند. آن ها وب سایت یا اپلیکیشن شرکت را مرور می کنند، بروشور خدمات را دانلود می کنند و برای دریافت اطلاعات بیشتر از طریق ایمیل یا تماس با روابط عمومی شرکت ارتباط برقرار می کنند. شرکت ها مایل هستند تا مشتریان بالقوه را به مشتری بالفعل (paid customer) تبدیل کنند؛ یعنی به افرادی که در عمل، خدمات یا محصولات آنان را خریداری می کنند. بدین منظور از تکنیک های متعددی در بازاریابی دیجیتال استفاده می شود؛ مانند: ارسال کد تخفیف و یا تماس و ارسال ایمیل. از آن جایی که پیاده سازی برخی از این روش ها برای شرکت ها هزینه بر است، شناسایی افرادی که به احتمال بیشتری ممکن است به یک مشتری بالفعل تبدیل شوند، برای شرکت ها حائز اهمیت است.

<br></br>

<br></br>

<br></br>

<br></br>

<br></br>

time_spent_on_website: Total time (seconds) spent on the website.
<br></br
page_views_per_visit: Average number of pages on the website viewed during the visits
<br></br>
last_activity: Last interaction between the potentional customer and XonLearn
<br></br>
Email Activity: Seeking details about the program through email, Representative shared information with a potentional customer like a brochure of the program, etc

<br></br>
status:Flag indicating whether the potentional customer was converted to a paid customer or not. The class 1 represents the paid customer and class 0 represents the unpaid customer

## Libraries and reproducibility | کتابخانه‌ها و بازتولیدپذیری

**English.** We use standard pandas, matplotlib/seaborn and scikit-learn tools. Randomness is fixed at 0; no `category_encoders` dependency is needed.

**فارسی.** از ابزارهای استاندارد pandas، matplotlib/seaborn و scikit-learn استفاده می‌کنیم. بذر تصادفی ۰ است و نیازی به `category_encoders` نیست.

```python
import warnings; warnings.filterwarnings('ignore')
from pathlib import Path
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix
RANDOM_STATE=0
sns.set_theme(style='whitegrid')
```

## Load and describe data | بارگذاری و توصیف داده

**English.** `ID` is a row identifier, not a predictive feature; it is explicitly removed from model inputs to prevent spurious patterns. The original table is retained for exploration.

**فارسی.** `ID` شناسه ردیف است، نه ویژگی پیش‌بین؛ برای جلوگیری از الگوهای کاذب صریحاً از ورودی مدل حذف می‌شود. جدول اصلی برای تحلیل اکتشافی حفظ می‌شود.

```python
data_path=Path('XonLearn.csv')
if not data_path.exists(): data_path=Path('/mnt/data/XonLearn.csv')
data=pd.read_csv(data_path)
print('Loaded:',data_path.resolve(),'| shape:',data.shape)
print(data.isna().sum().to_string())
display(data.head())
display(data.describe(include='all').T)
```

## Exploratory analysis | تحلیل اکتشافی

**English.** These descriptive plots summarize class balance and selected associations; they do not establish causality.

**فارسی.** نمودارهای توصیفی توازن کلاس و چند رابطه را خلاصه می‌کنند؛ علیت را اثبات نمی‌کنند.

```python
fig,ax=plt.subplots(1,2,figsize=(12,4))
sns.countplot(data=data,x='status',ax=ax[0]); ax[0].set_title('Target counts (status)')
sns.countplot(data=data,x='profile_completed',hue='status',ax=ax[1]); ax[1].set_title('Profile completion by status')
plt.tight_layout(); plt.show()
print('Target proportions:',data.status.value_counts(normalize=True).sort_index().round(3).to_dict())
fig,ax=plt.subplots(1,2,figsize=(12,4))
sns.countplot(data=data,x='first_interaction',hue='status',ax=ax[0]); ax[0].tick_params(axis='x',rotation=15); ax[0].set_title('First interaction by status')
sns.boxplot(data=data,x='status',y='time_spent_on_website',ax=ax[1]); ax[1].set_title('Website time by status')
plt.tight_layout(); plt.show()
numeric=data.select_dtypes(include='number').drop(columns=['ID'],errors='ignore')
plt.figure(figsize=(9,6)); sns.heatmap(numeric.corr(),annot=True,fmt='.2f',cmap='vlag',center=0); plt.title('Numeric correlations'); plt.tight_layout(); plt.show()
```

## Modeling setup and split | آماده‌سازی مدل و تقسیم داده

**English.** A single split uses `test_size=0.30`, `random_state=0`, and `stratify=y`; stratification preserves class proportions in this imbalanced target. Categorical variables are one-hot encoded with `handle_unknown='ignore'` (no false category order), while numeric values are median-imputed. Both steps are inside each model Pipeline and are fitted on training data only, preventing preprocessing leakage. Uppercase `ID` is dropped explicitly.

**فارسی.** یک تقسیم با `test_size=0.30`، `random_state=0` و `stratify=y` انجام می‌شود؛ لایه‌بندی نسبت کلاس‌ها را در هدف نامتوازن حفظ می‌کند. متغیرهای طبقه‌ای با `OneHotEncoder(handle_unknown='ignore')` کدگذاری می‌شوند تا ترتیب کاذب ایجاد نشود و مقادیر گمشده عددی با میانه جایگزین می‌شوند. هر دو مرحله داخل Pipeline هستند و فقط روی آموزش برازش می‌شوند تا نشت اطلاعات رخ ندهد. ستون بزرگ‌نویس `ID` صریحاً حذف می‌شود.

```python
assert 'ID' in data and 'status' in data
X=data.drop(columns=['ID','status']); y=data['status']
assert 'ID' not in X
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.30,random_state=RANDOM_STATE,stratify=y)
num_cols=X.select_dtypes(include='number').columns.tolist(); cat_cols=X.select_dtypes(exclude='number').columns.tolist()
def make_preprocessor():
    return ColumnTransformer([('num',SimpleImputer(strategy='median'),num_cols),
      ('cat',Pipeline([('impute',SimpleImputer(strategy='most_frequent')),('onehot',OneHotEncoder(handle_unknown='ignore'))]),cat_cols)])
print('ID excluded:', 'ID' not in X.columns, '| train/test:',X_train.shape,X_test.shape)
print('Train proportions:',y_train.value_counts(normalize=True).sort_index().round(3).to_dict())
print('Test proportions:',y_test.value_counts(normalize=True).sort_index().round(3).to_dict())
```

## Model comparison | مقایسه مدل‌ها

**English.** Decision Tree and Random Forest are evaluated consistently on the same holdout set using accuracy, class-1 precision, recall, F1, classification report and confusion matrix. Accuracy alone may mislead under imbalance.

**فارسی.** درخت تصمیم و جنگل تصادفی روی یک مجموعه آزمون مشترک با accuracy، precision کلاس ۱، recall، F1، گزارش طبقه‌بندی و ماتریس درهم‌ریختگی ارزیابی می‌شوند. در داده نامتوازن accuracy به‌تنهایی می‌تواند گمراه‌کننده باشد.

```python
models={'Decision Tree':Pipeline([('preprocess',make_preprocessor()),('model',DecisionTreeClassifier(max_depth=7,random_state=0))]),
'Random Forest':Pipeline([('preprocess',make_preprocessor()),('model',RandomForestClassifier(n_estimators=300,random_state=0,n_jobs=-1))])}
rows=[]; predictions={}
for name,model in models.items():
    model.fit(X_train,y_train); pred=model.predict(X_test); predictions[name]=pred
    rows.append({'Model':name,'Accuracy':accuracy_score(y_test,pred),'Precision (class 1)':precision_score(y_test,pred,zero_division=0),'Recall (class 1)':recall_score(y_test,pred,zero_division=0),'F1 (class 1)':f1_score(y_test,pred,zero_division=0)})
    print('\n'+name+' classification report')
    print(classification_report(y_test,pred,labels=[0,1],target_names=['Not converted (0)','Converted (1)'],zero_division=0))
    cm=confusion_matrix(y_test,pred,labels=[0,1]); print('Confusion matrix [actual rows, predicted columns]:\n',cm)
    sns.heatmap(cm,annot=True,fmt='d',cmap='Blues',xticklabels=['Pred 0','Pred 1'],yticklabels=['Actual 0','Actual 1']); plt.title(name+' confusion matrix'); plt.tight_layout(); plt.show()
results_df=pd.DataFrame(rows).set_index('Model'); display(results_df.style.format('{:.3f}'))
```

## Tree view and feature importance | نمایش درخت و اهمیت ویژگی

**English.** The fitted tree is shown to depth 3 for readability. Random-forest one-hot importances are aggregated back to original columns. Importance is predictive association, not causation, and can be unstable with correlated features.

**فارسی.** درخت برازش‌شده برای خوانایی تا عمق ۳ نمایش داده می‌شود. اهمیت one-hot جنگل تصادفی به ستون‌های اصلی جمع می‌شود. اهمیت، وابستگی پیش‌بینانه است نه علیت و در ویژگی‌های هم‌بسته ممکن است ناپایدار باشد.

```python
dt=models['Decision Tree']; names=dt.named_steps['preprocess'].get_feature_names_out()
plt.figure(figsize=(20,9)); plot_tree(dt.named_steps['model'],feature_names=names,class_names=['Not converted','Converted'],filled=True,max_depth=3,fontsize=8); plt.tight_layout(); plt.show()
rf=models['Random Forest']; enames=rf.named_steps['preprocess'].get_feature_names_out(); vals=rf.named_steps['model'].feature_importances_
agg={c:0. for c in X.columns}
for ename,val in zip(enames,vals):
    clean=ename.split('__',1)[-1]
    match=next((col for col in X.columns if clean==col or clean.startswith(col+'_')),None)
    if match is not None: agg[match]+=float(val)
importance=pd.Series(agg,name='Aggregated importance').sort_values(ascending=False)
display(importance.to_frame().style.format('{:.4f}'))
importance.sort_values().plot.barh(figsize=(9,5),color='teal'); plt.xlabel('Summed impurity-based importance'); plt.title('Random Forest importance by original feature'); plt.tight_layout(); plt.show()
```

## Hyperparameter exploration | کاوش ابرپارامتر

**English.** A small `GridSearchCV` tests tree depth with stratified five-fold CV on training data only, optimizing F1; preprocessing is refit inside every fold. This is genuine CV, not an exhaustive search. The final test set remains held out.

**فارسی.** `GridSearchCV` محدود، عمق درخت را با اعتبارسنجی متقاطع لایه‌ای پنج‌تایی و فقط روی آموزش می‌سنجد و F1 را بهینه می‌کند؛ پیش‌پردازش در هر فولد دوباره برازش می‌شود. این اعتبارسنجی متقاطع واقعی اما جست‌وجویی محدود است. آزمون نهایی کنار گذاشته می‌شود.

```python
cv=StratifiedKFold(n_splits=5,shuffle=True,random_state=0)
search=GridSearchCV(Pipeline([('preprocess',make_preprocessor()),('model',DecisionTreeClassifier(random_state=0))]),
 {'model__max_depth':[3,5,7,10,None]},scoring='f1',cv=cv,n_jobs=-1)
search.fit(X_train,y_train)
print('Best CV params:',search.best_params_,'| mean CV F1:',round(search.best_score_,4))
tuned=search.predict(X_test)
tuned_row={'Model':'Tuned Decision Tree (5-fold CV)','Accuracy':accuracy_score(y_test,tuned),'Precision (class 1)':precision_score(y_test,tuned,zero_division=0),'Recall (class 1)':recall_score(y_test,tuned,zero_division=0),'F1 (class 1)':f1_score(y_test,tuned,zero_division=0)}
print(classification_report(y_test,tuned,labels=[0,1],target_names=['Not converted (0)','Converted (1)'],zero_division=0))
print('Tuned tree confusion matrix:\n',confusion_matrix(y_test,tuned,labels=[0,1]))
```

## Final results, interpretation and limitations | نتایج، تفسیر و محدودیت‌ها

**English.** The results table is generated from actual test predictions. The comparison statement is computed from actual F1 values, not assumed.
